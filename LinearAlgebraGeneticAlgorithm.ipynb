```python
# --- Create Initial Population ---

print("Generating initial ship population...")
initial_population_size = 10000

# Generate all ships first by calling the genShips function
all_generated_ships = []
for _ in range(initial_population_size):
  all_generated_ships.append(genShips())

print(f"Generated {len(all_generated_ships)} ships, now checking and filtering them.")

# --- Filter Population ---

myShips = [] # Final list of valid, weaponized ships
filteredNoWeaponShips = [] # Valid ships missing a weapon
invalidShips = [] # Ships that fail validation

for ship in all_generated_ships:
  vector = ship.getVector()

  # 1. Check for general validity
  # This check is still useful in case baseShip + Hyperdrive is invalid
  is_valid = (vector[8] >= 200 and vector[12] >= 0 and vector[13] >= 0 and
              vector[14] >= 0 and vector[22] >= 0 and vector[28] < 50)

  # 2. Check if the ship has any weapon outfit
  has_weapon = False
  if is_valid: # No need to check for weapons if it's already invalid
    for outfit in ship.getOutfits():
      if outfit in weapon_outfits:
        has_weapon = True
        break

  # 3. Categorize the ship
  if is_valid and has_weapon:
    myShips.append(ship)
  elif not is_valid:
    invalidShips.append(ship)
  elif is_valid and not has_weapon:
    # This catches ships where genShips failed to add a weapon in its attempts
    filteredNoWeaponShips.append(ship)

print("Ship checking done, here is the final list of ships in initial population")
print(f"Number of valid and weaponized ships: {len(myShips)}")
print(f"Number of invalid ships (removed): {len(invalidShips)}")
print(f"Number of valid ships without weapons (removed): {len(filteredNoWeaponShips)}")
```
